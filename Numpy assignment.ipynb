```python
# generate report card with total, average, percentage and grade
import numpy as np
marks=np.array([80,75,90,85,70])
total=np.sum(marks)
average=np.mean(marks)
percentage=total/5
if percentage>=90:
    grade="A"
elif percentage>=75:
    grade="B"
elif percentage>=60:
    grade="C"
elif percentage>=35:
    grade="D"
else:
    grade="F"
print("Marks:",marks)
print("Total:",total)
print("Average:",average)
print("Percentage:",percentage)
print("Grade:",grade)
```
