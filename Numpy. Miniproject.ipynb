```python
# generate marks for 50 students
import numpy as np
marks=np.random.randint(0,101,50)
print(marks)
```

```python
# calculate total, average highest,lowest
import numpy as np
marks=np.array([78,45,92,61,35])
print("Total:",np.sum(marks))
print("Average:",np.mean(marks))
print("Highest:",np.max(marks))
print("Lowest:",np.min(marks))
```

```python
# count above 75,below 35, between 35-75
import numpy as np
marks=np.array([78,45,92,61,35,84,20,75])
print("Above 75:",np.sum(marks>75))
print("Below 35:",np.sum(marks<35))
print("Between 35-75:",np.sum((marks>=35)&(marks<=75)))
```

```python
# assign grades
import numpy as np
marks=np.array([95,82,74,66,58,40])
grades=np.where(marks>=90,'A',
       np.where(marks>=75,'B',
       np.where(marks>=60,'C',
       np.where(marks>=35,'D',
'F'))))
print("Marks:",marks)
print("Grades:",grades)
```

```python
# display top 5 marks
import numpy as np
marks=np.array([78,45,39,20,65,70,98,23])
top5=np.sort(marks)[-5:][::-1]
print(top5)
```

```python
# display bottom 5 marks
import numpy as np
marks=np.array([87,78,45,39,20,65,70,98,23])
bottom5=np.sort(marks)[:5]
print(bottom5)
```

```python
# find median and standard deviation
import numpy as np
marks=np.array([87,78,45,39,20,65,70,98,23])
print("Median:",np.median(marks))
print("Standard Deviation:",np.std(marks))
```

```python
# create 5-subject marks matrix and calculate totals
import numpy as np
marks=np.array([[80,75,90,85,70],[60,65,70,75,80],[90,95,85,80,88]])
print("Marks Matrix:")
print(marks)
total=np.sum(marks,axis=1)
print("Total Marks:",total)
```

```python
# find topper in each subject
import numpy as np
marks=np.array([[80,75,90,85,70],[60,65,70,75,80],[90,95,85,80,88]])
topper=np.max(marks,axis=0)
print("Top Marks in Each Subject:")
print(topper)
```

```python
# generate report card with total, average, percentage and grade
import numpy as np
marks=np.array([80,75,90,85,70])
total=np.sum(marks)
average=np.mean(marks)
percentage=total/5
if percentage>=90:
    grade="A"
elif percentage>=75:
    grade="B"
elif percentage>=60:
    grade="C"
elif percentage>=35:
    grade="D"
else:
    grade="F"
print("Marks:",marks)
print("Total:",total)
print("Average:",average)
print("Percentage:",percentage)
print("Grade:",grade)
```
