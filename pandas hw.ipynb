```python
# create a 1D series
import pandas as pd
s=pd.Series([10,20,30,40,50])
print(s)
```

```python
# addition
import pandas as pd
s=pd.Series([10,20,30])
print(s+5)
```

```python
# substraction
import pandas as pd
s=pd.Series([10,20,30])
print(s-2)
```

```python
# multiplication
import pandas as pd
s=pd.Series([10,20,30,40])
print(s*3)
```

```python
# division
import pandas as pd
s=pd.Series([10,20,30,40])
print(s/4)
```

```python
# modules
import pandas as pd
s=pd.Series([10,20,30,40])
print(s%4)
```

```python
# power
import pandas as pd
s=pd.Series([10,20,30,40])
print(s**2)
```

```python
# greater than 
import pandas as pd
s=pd.Series([10,20,30,40])
print(s>30)
```

```python
#  less than
import pandas as pd
s=pd.Series([10,20,30,40])
print(s<20)
```

```python
# equal than
import pandas as pd
s=pd.Series([10,20,30,40])
print(s==30)
```
