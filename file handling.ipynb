```python
# count the number of lines,wordsj, and characters in a text file
file=open("Demo.txt","r")
data=file.read()
lines=data.split("\n")
words=data.split()
print("Lines:",len(lines))
print("Words:",len(words))
print("Characters:",len(data))
file.close()
```

```python
# take user input and store it until the user types"exit"
file=open("Data.txt","w")
while True:
    text=input("Enter text:")
    if text.lower()== "exit":
        break
    file.write(text+"\n")
file.close()
print("Data saved successfully.")
```
