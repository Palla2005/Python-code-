```python
# count words in a sentence
string=input("Enter a sentence:")
count=1
for ch in string:
    if ch=="":
        count+=1
print("Words:",count)
```

```python
# count alphabets, digits and special characters
string=input("Enter a string:")
a=d=sp=0
for ch in string:
    if ch.isalpha():
        a+=1
    elif ch.isdigit():
        d+=1
    else:
        sp+=1
print("Alphabets:",a)
print("Digits:",d)
print("Special:",sp)
```

```python
# Remove all spaces
string=input("Enter a string:")
result=""
for ch in string:
    if ch !="":
        result+=ch
print(result)
```

```python
#Replace vowels with *
string=input("Enter a string:")
result=""
for ch in string:
    if ch.lower() in "aeiou":
        result+="*"
    else:
        result+=ch
print(result)
```

```python
# digits and find sum
string=input("Enter a string:")
sum=0
for ch in string:
    if ch.isdigit():
        sum+=int(ch)
print("Sum:",sum)
```

```python
# convert to uppercase without.upper()
string=input("Enter a string:")
result=""
for ch in string:
    if'a'<=ch<='z':
        result+=chr(ord(ch)-32)
    else:
        result+=ch
print(result)
```
