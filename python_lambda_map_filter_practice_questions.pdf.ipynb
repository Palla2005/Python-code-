```python
# lambda
```

```python
#add two numbers
add=lambda a,b:a+b
print(add(10,20))
```

```python
# subtract two numbers
sub=lambda a,b:a-b
print(sub(50,39))
```

```python
# multiply two numbers
mul=lambda a,b:a*b
print(mul(4,5))
```

```python
# Divide two numbers
div=lambda a,b:a/b
print(div(20,8))
```

```python
#Find square
square=lambda x:x**2
print(square(6))
```

```python
# find cube
cube=lambda x:x**3
print(cube(7))
```

```python
# check even or odd
even=lambda x: "Even" if x % 2==0 else "Odd"
print(even(7))
```

```python
# greater of two
greater=lambda a,b:a if a>b else b
print(greater(15,20))
```

```python
# largest among three
largest=lambda a,b,c:max(a,b,c)
print(largest(10,28,15))
```

```python
# sort tuples by second element
data=[(1,3),(2,1),(4,2)]
print(sorted(data, key=lambda x:x[1]))
```

```python
# map function
```

```python
#use map() to square every number
nums=[1,2,3,4]
print(list(map(lambda x:x*x,nums)))
```

```python
#cube every number
nums=[1,2,3]
print(list(map(lambda x:x**3,nums)))
```

```python
# convert string to uppercase
names=["pallavi","Isha"]
print(list(map(str.upper,names)))
```

```python
#convert strings to lowercase
names=["RAM","PALLAVI"]
print(list(map(str.lower,names)))
```

```python
#add 10 to every element
nums=[10,20,30]
print(list(map(lambda x:x+10,nums)))
```

```python
#convert numetic strings to integers
nums=["1","2","3"]
print(list(map(int,nums)))
```

```python
#find length of each word
words=["cat","apple","python"]
print(list(map(len,words)))
```

```python
# convert celsius to fahrenheit
c=[0,20,30]
print(list(map(lambda x:(x*9/5)+32,c)))
```

```python
#multiply corresponding elements of two lists
a=[1,2,3]
b=[4,5,6]
print(list(map(lambda x,y:x*y,a,b)))
```

```python
# calculate 10% discont for each product price
price=[100,200,300]
print(list(map(lambda x:x*0.9,price)))
```

```python
# filter function
```

```python
#filter even numbers
nums=[1,2,3,4,5,6]
print(list(filter(lambda x:x%2==0,nums)))
```

```python
#filter odd numbers
nums=[1,2,3,4,5]
print(list(filter(lambda x:x%2!=0,nums)))
```

```python
#filter numbers greater than 50
nums=[20,60,80,40]
print(list(filter(lambda x:x>50,nums)))
```

```python
#filter positive numbers
nums=[-2,3,-1,5]
print(list(filter(lambda x:x>0,nums)))
```

```python
# filter strings longer than 5 characters
words=["cat","python","apple","go"]
print(list(filter(lambda x:len(x)>5,words)))
```

```python
# filter students scorting above 75
marks=[60,80,90,70]
print(list(filter(lambda x:x>75,marks)))
```

```python
#filter numbers divisible by both 3 and 5
nums=[15,30,10,9,45]
print(list(filter(lambda x:x%3==0 and x%5==0,nums)))
```

```python
#filter numbers divisible by both 3 and 5
nums=[1,2,3,4,5]
result=list(map(lambda x:x*x,filter(lambda x:x%2==0,nums)))
print(result)
```

```python
#filter negative numbers
nums=[-2,3,-1,5]
print(list(filter(lambda x:x<0,nums)))
```
