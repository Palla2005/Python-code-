```python
# proft or loss
cost=float(input("Enter cost price:"))
sell=float(input("Enter selling price:"))
if sell>cost:
    print("Profit=",sell-cost)
elif sell<cost:
    print("Loss=",cost-sell)
else:
    print("No Profit No Loss")
```

```python
# salary bonus
salary=float(input("Enter salary:"))
exp=int(input("Enter years of experience:"))
if exp>5:
    bonus=salary*0.10
    print("Bonus=",bonus)
    print("New salary=",salary+bonus)
else:
    print("No Bonus")
```

```python
# bill with discount
p1=float(input("Price 1:"))
p2=float(input("Price 2:"))
p3=float(input("Price 3:"))
p4=float(input("Price 4:"))
total=p1+p2+p3+p4
if total>5000:
    discount=total*0.10
else:
    discount=total*0.05
print("Total Bill=",total)
print("Discount=",discount)
print("Final Bill=",total-discount)
```

```python
# attendance percentage
held=int(input("Enter classes held:"))
attended=int(input("Enter classes attended:"))
percentage=(attended/held)*100
print("Attendance=",percentage,"%")
if percentage>=75:
    print("Allowed to sit in exam")
else:
    print("Not allowed to sit in exam")
```

```python
# electricity bill
units=int(input("Enter electricity units:"))
if units<=100:
    bill=0
elif units<=200:
    bill=(units-100)*5
else:
    bill=(100*5)+(units-200)*10
print("Electricity Bill=Rs.",bill)
```

```python
# maximum of four numbers(nested if)
a=int(input("Enter first number:"))
b=int(input("Enter second number:"))
c=int(input("Enter thirs number:"))
d=int(input("Enter forth number:"))
if a>b:
    if a>c:
        if a>d:
            print("Maximum=",a)
        else:
            print("Maximum=",d)
    else:
        if c>d:
            print("Maximum=",c)
        else:
            print("Maximum=",d)
else:
    if b>c:
        if b>d:
            print("Maximum=",b)
        else:
            print("Maximum=",d)
    else:
        if c>d:
            print("Maximum=",c)
        else:
            print("Maximum=",d)
    
```
