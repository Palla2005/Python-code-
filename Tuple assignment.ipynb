```python
# create a tuple with 5 numbers and print all elements
tuple=(10,20,30,40,50)
print(tuple)
```

```python
# print first and last element using indexing
tuple=(10,20,30,40,50)
print(tuple[0])
print(tuple[-1])
```

```python
# find length of a tuple
tuple=(10,20,30,40,50)
print(len(tuple))
```

```python
#count how many timews a number appears
tuple=(10,20,10,30,10)
print(tuple.count(10))
```

```python
# find maximum and minimum value
tuple1=(10,20,5,40,15)
print(max(tuple1))
print(min(tuple1))
```

```python
# join two tuples
tuple1=(1,2,3)
tuple2=(4,5,6)
print(tuple1+tuple2)
```

```python
# repeat tuple 3 times
tuple=(1,2)
print(tuple*3)
```

```python
# take 5 numbers from user
numbers=()
for i in range(5):
    num=int(input("Enter a number:"))
    numbers+=(num,)
print("Tuple:",numbers)
```

```python
# print tuple using loop
tuple=(10,20,30)
for i in tuple:
    print(i)
```

```python
# reverse a tuple using slicing
tuple=(10,20,30,40)
print(tuple[::-1])
```

```python
# check elemente exists or not
tuple=(10,20,30)
print(20 in tuple)
```

```python
# print elements at even index positions
tuple=(10,20,30,40,50)
print(tuple[::2])
```

```python
# unpack tuple into variables
tuple=(10,20,30)
a,b,c=tuple
print(a,b,c)
```

```python
#swap two numbers using tuple
a=10
b=20
a,b=b,a
print(a,b)
```

```python
# function returns multiple values
def data():
    return(10,20)
print(data())
```

```python
#store student data
student=(1,"Rahul", 85)
print(student)
```

```python
# calculate total bill using sum()
prices=(100,200,300)
print(sum(prices))
```

```python
# find average temperature
temp=(30,32,31,29,33,34,30)
avg=sum(temp)/len(temp)
print(avg)
```

```python
# nested tuple of students
students=((1,"Rahul"),(2,"Priya"))
for s in students:
    print(s)
```
