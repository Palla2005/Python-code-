<h1 style="font-size:50px:">Feature Engineering</h1>

```python
# price category
data["Price_Category"] = np.where(data["Price"] > 100,"Expensive","Affordable")
data[["Price", "Price_Category"]].head()
```

```python
# rating category
data["Rating_Category"]=np.where(data["Rating"]>=4,"Good","Average")
data[["Rating", "Rating_Category"]].head()
```

```python
# revenue feature
data["Revenue"]=data["Price"]*data["Availability"]
data[["Price","Availability","Revenue"]].head()
```

```python
# price rating
data["Price_per_Rating"]=data["Price"]/data["Rating"]
data[["Price","Rating","Price_per_Rating"]].head()
```

```python
# product name length
data["Product_Name_Length"]=data["Product_Name"].str.len()
data[["Product_Name","Product_Name_Length"]].head()
```

```python
# brand length
data["Brand_Length"]=data["Brand"].str.len()
data[["Brand","Brand_Length"]].head()
```

```python
# category encoding
data["Category_Code"]=data["Category"].astype("category").cat.codes
data[["Category","Category_Code"]].head()
```

```python
# high rating feature
data["High_Rating"]=np.where(data["Rating"]>= 4.5,1,0)
data[["Rating","High_Rating"]].head()
```
