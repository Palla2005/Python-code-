```python
# create a dictionary of 5 students and print it
students={"Amit":80,
          "Neha":90,
          "Rahul":75,
          "Pooja":88,
          "Kiran":95}
print(students)
```

```python
# access value using key
student={"Amit":80,"Riya":85}
print(student["Amit"])
```

```python
# add a new key-value pair
student={"Amit":80,"Riya":85}
student["Neha"]=90
print(student)
```

```python
# update a value
s={"Amit":80,"Riya":85}
s["Amit"]=95
print(s)
```

```python
# delete a key
s={"Amit":80,"Riya":85}
del s["Riya"]
print(s)
```

```python
# print keys and values using loop
s={"Amit":80,"Riya":85}
for key,value in s.items():
    print(key,value)
```

```python
# create a set and print elements
s={10,20,30,40}
print(s)
```

```python
# add elments to a set
s={10,20,30}
s.add(40)
print(s)
```

```python
# remove element from a set
s={10,20,30,40}
s.remove(20)
print(s)
```

```python
# count frequency of characters
text="apple"
freq={}
for ch in text:
    freq[ch]=freq.get(ch,0)+1
print(freq)
```

```python
# Merge two dictionaries
d1={"A":1,"B":2}
d2={"C":3,"D":4}
d1.update(d2)
print(d1)
```

```python
# check if key exists
s={"Amit":80,"Riya":85}
if "Amit" in s:
    print("Key Found")
else:
    print("Key Not Found")
```

```python
# find union of two sets
A={1,2,3}
B={3,4,5}
print(A|B)
```

```python
# find intersection of two sets
A={1,2,3}
B={3,4,5}
print(A|B)
```

```python
#find difference between two sets
A={1,2,3}
B={3,4,5}
print(A-B)
```

```python
# convert list into set and remove duplicates
st1=[1,2,2,3,4,4,5]
s=set(st1)
print(s)
```
