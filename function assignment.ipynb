```python
# welcome message
def welcome():
    print("Welcome to Python!")
welcome()
```

```python
# accept student name
def student(name):
    print("Student Name:",name)
student("Pallavi")
```

```python
# sum of two numbers
def add(a,b):
    return a+b
print(add(10,20))
```

```python
# square of a number
def square(n):
    return n*n
print(square(6))
```

```python
# even or odd
def check(n):
    if n%2==0:
        print("Even")
    else:
        print("Odd")
check(7)
```

```python
# largest among three numbers
def largest(a,b,c):
    print(max(a,b,c))
largest(10,50,30)
```

```python
# factorial
def fact(n):
    f=1
    for i in range(1,n+1):
        f*=i
    return f
print(fact(5))
```

```python
# average of a list
def average(st1):
    return sum(st1)/len(st1)
print(average([10,20,30,40]))
```

```python
# count vowels
def vowels(s):
    count=0
    for i in s.lower():
        if i in "aeiou":
            count+=1
    return count
print(vowels("Python"))
```

```python
# quotient and remainder
def divide(a,b):
    return a//b,a%b
q,r=divide(17,5)
print("Quotient:",q)
print("Remainder:",r)
```

```python
# area of rectange using positional arguments
def area(length, width):
    print("Area=",length*width)
area(10,5)
```

```python
# display employee details using keyword arguments
def employee(name, age, salary):
    print("Name:",name)
    print("Age:",age)
    print("Salary:",salary)
employee(name="Pallavi",age=20,salary=25000)
```

```python
#simple interest using defalt arguments
def si(p, r=5,t=2):
    print("Simple Interest=",(p*r*t)/100)
si(10000)
```

```python
# find the sum using *args
def total(*args):
    print("Sum=",sum(args))
total(10,20,30,40)
```

```python
# display name, age and city sing keyword arguments
def details(name, age, city):
    print("Name:",name)
    print("Age:",age)
    print("City:",city)
details(name="Pallavi",age=20,city="Pune")
```

```python
# calculator using functions
def add(a,b):
    return a+b
def sub(a,b):
    return a-b
print("Addition=",add(10,5))
print("Subtraction=",sub(10,5))
```

```python
# menu-driven calculator
def add(a,b):
    return a+b
def sub(a,b):
    return a-b
choice=1
a=20
b=10
if choice==1:
    print("Addition=",add(a,b))
else:
    print("Subtraction=",sub(a,b))
```

```python
# check prime numbers
def prime(n):
    if n>1:
        for i in range(2,n):
            if n%i==0:
                return "Not Prime"
        return "Prime"
print(prime(7))
```

```python
# fibonacci series
def fib(n):
    a,b=0,1
    for i in range(n):
        print(a,end="")
        a, b=b, a+b
fib(5)
```

```python
# check string is palindrome
def palindrome(s):
    if s==s[::-1]:
        print("Palindrome")
    else:
        print("Not Palindrome")
palindrome("madam")
```

```python
# count uppercase, lowercase, digits and special characters
s="PyThon@123"
u=l=d=sp=0
for ch in s:
    if ch.isupper():
        u+=1
    elif ch.islower():
        l+=1
    elif ch.isdigit():
        d+=1
    else:
        sp+=1
print("Uppercase:",u)
print("Lowercase:",l)
print("Digits:",d)
print("Special:",sp)
```

```python
# accept multiple numbers using *args and return maximum
def maximum(*args):
    return max(args)
print("Maximum=",maximum(10,50,30,80,40))
```
