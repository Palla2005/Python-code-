```python
# count the number of digits in a given number
num=int(input("Enter a number:"))
count=0
while num>0:
    count+=1
    num//=10
print("Digits=",count)
```

```python
# reverse a given number using a loop
num=int(input("Enter a number:"))
rev=0
while num>0:
    digit=num%10
    rev=rev*10+ digit
    num//=10
    print("Reverse=",rev)
```

```python
# find the factorial of a number using a loop
num=int(input("Enter a number:"))
fact=1
for i in range(1,num+1):
    fact*=i
print("factorial=",fact)
```

```python
# stop a loop when the value reaches 5 using break
for i in range(1,11):
    if i ==5:
        break
    print(i)
```

```python
# skip the number 5 in a loop using continue
for i in range(1,11):
    if i==5:
        print(i)
```

```python
# that uses pass inside a loop
for i in range(1,6):
    if i==3:
       pass
    print(i)
```
