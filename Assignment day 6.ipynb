```python
# a while loop to print number from 1 to 10
i=1
while i<=10:
     print(i)
     i+=1
```

```python
# a while loop to print even number from 2 to 20
i=2
while i<=20:
    print(i)
    i+=2
```

```python
# a while loop to calculate the sum of numbers from 1 to 10
i=1
sum=0
while i<=10:
    sum+=1
    i+=1
print("Sum=",sum)
```

```python
# a while loop to print the multiplication table of 5
i=1
while i<=10:
    print("5X",i,"=",5*i)
    i+=1
```

```python
# nested loop program to print *,**,***,****
for i in range(1,5):
    for j in range(i):
        print("*",end="")
    print()
```

```python
# a nested loop program to print 1,12,123,1234
for i in range(1,5):
    for j in range(1,i+1):
        print(j,end="")
    print()
```

```python
# a while loop to find the factorial of a number entered by the user
num=int(input("Enter a number:"))
fact=1
while i<=num:
    fact*=i
    i+=1
print("factorial=",fact)
```
