```python
# count digits,letters and spaces
s="Python 123"
d=l=sp=0
for ch in s:
    if ch.isdigit():
        d+=1
    elif ch.isalpha():
        l+=1
    elif ch.isspace():
        sp+=1
print("Digits=",d)
print("Letter=",l)
print("Spaces=",sp)       
```
