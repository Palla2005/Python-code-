```python
# create a list of 5 numbers and print all elements
numbers=[10,20,30,40,50]
for i in numbers:
    print(i)
```

```python
# take 5 numbers from user,store in list, and print sum.
numbers=[]
for i in range(5):
      num=int(input("Enter number:"))
      numbers.append(num)
print("Sum=",sum(numbers))
```

```python
# find maximum and minimum from a list
numbers=[12,45,7,89,23]
maximum=max(numbers)
minimum=min(numbers)
print("List:",numbers)
print("Maximum number:",maximum)
print("Minimum number:",minimum)
```

```python
# reverse a list using slicing
numbers=[1,2,3,4,5]
print("Reversed list:",numbers[::-1])
```

```python
# check if an exists in a list
numbers=[10,20,30,40,50]
x=30
if x in numbers:
    print("Element exists")
else:
    print("Element does not exist")
```

```python
# count occurrence of an element in list
numbers=[10,20,30,40,50]
print("count=",numbers.count(10))
```

```python
#sort list in ascending and descending order
numbers=[50,30,40,20,10]
numbers.sort()
print("Ascending:",numbers)
numbers.sort(reverse=True)
print("Decending:",numbers)
```
