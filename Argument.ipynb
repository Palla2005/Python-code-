```python
#positional argument substract numbers with value 10 and 5
def sub(a,b):
   print("substraction:",a-b)
sub(10,5)
```

```python
# positional argument display student name and age 
def student(name,age):
    print("Name:",name)
    print("Age:",age)
student("Pallavi",18)
```

```python
# positional argument multiply three numbers
def mul(a,b,c):
    print("multiplication:",a*b*c)
mul(5,9,3)
```
