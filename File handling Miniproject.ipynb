```python
# create a student record system that allows adding student data(name,roll, marks) and saving the data to a file
f=open("students.txt","a")
name=input("Enter a Name:")
roll=input("Enter Roll No:")
marks=input("Enter marks:")
f.write(name+","+roll+","+marks+"\n")
f.close()
print("Student record added successfully.")
```

```python
# search student by roll number
roll=input("Enter Roll No:")
f=open("students.txt","r")
found=False
for line in f:
    data=line.strip().split(",")
    if data[1]==roll:
        print("Record Found:",line)
        found=True
if not found:
    print("Record Not Found")
f.close()
```

```python
#update student marks
roll=input("Enter Roll No:")
new_marks=input("Enter New Marks:")
f=open("students.txt","r")
lines=f.readlines()
f.close()
f=open("students.txt","w")
for line in lines:
    data=line.strip().split(",")
    if data[1]==roll:
        f.write(data[0]+","+data[1]+","+new_marks+"\n")
    else:
        f.write(line)
f.close()
print("Marks Updated Successfully")
```

```python
#Delete student record
roll=input("Enter Roll No:")
f=open("students.txt","r")
lines=f.readlines()
f.close()
f=open("students.txt","w")
for line in lines:
    data=line.strip().split(",")
    if data[1]!=roll:
        f.write(line)
f.close()
print("Record Deleted Successfully")
```

```python
# Student Record System - Add and Save Data
file = open("students.txt", "a")
n = int(input("Enter number of students: "))
for i in range(n):
    roll = input("Enter Roll No: ")
    name = input("Enter Name: ")
    marks = input("Enter Marks: ")
    file.write(roll + "," + name + "," + marks + "\n")
file.close()
print("Student record saved successfully.")
```

```python
# display students with markds greater than 75
f=open("students.txt","r")
for line in f:
    data=line.strip().split(",")
    if int(data[2])>75:
        print(line)
f.close()
```

```python
# export top 3 students to a new file
students=[]
file=open("students.txt","r")
for line in file:
    roll,name,marks=line.strip().split(",")
    students.append([roll,name,int(marks)])
file.close()
students.sort(key=lambda x:x[2],reverse=True)
new_file=open("top3students.txt","w")
for student in students[:3]:
    new_file.write(f"{student[0]},{student[1]},{student[2]}\n")
new_file.close()
print("Top 3 students explorted to top3students.txt")
```

```python
# backup student data
source=open("students.txt","r")
data=source.read()
source.close()
backup=open("backup.txt","w")
backup.write(data)
backup.close()
print("Backup Created Successfully")
```
