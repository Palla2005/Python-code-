```python
# takes two number and an opetator(+,-,*,/)input and performs the opetation using if-elif-else
a=float(input("Enter first number:"))
b=float(input("Enter second number:"))
op=input("Enter opetator(+,-,*,/):")
if op=="+":
    print("Result=",a+b)
elif op=="-":
    print("Result=",a-b)
elif op=="*":
    print("Result=",a*b)
elif op=="/":
    if b!=0:
       print("Result=",a/b)
    else:
       print("Division by zero not allowed")
else:
       print("Invalid operator")
```

```python
# categorize a personals bmt using if-elif-else underflow, normal, overweight,abese
weight=float(input("Enter wight(kg):"))
height=float(input("Enter height(m):"))
bmi=weight/(height*height)
if bmi<18.5:
    print("Underflow")
elif bmi<25:
    print("Normal")
elif bmi<30:
    print("Overweight")
else:
    print("abses")
```

```python
# check if three sides from a valid triangle. valid,classigy it using nested if-else,Equilateral,Isosceles,scalene
a=int(input("Enter side1:"))
b=int(input("Enter side2:"))
c=int(input("Enter side3:"))
if a+b>c and a+c>b and b+c>a:
    if a==b==c:
        print("Equilateral Triangle")
    elif a==b or b==c or a==c:
        print("Isosceles Triangle")
    else:
        print("Scalene Triangle")
else:
       print("Invalid Triangle")
```

```python
# a simple ATM system using if-elif-else and nested if-else
pin=int(input("Enter PIN:"))
balance=10000
if pin==1234:
    amount=int(input("Enter withdrawal amount: "))
    if amount<=balance:
        balance=balance-amount
        print("Remaining Balance=",balance)
    else:
       print("Insufficient Balance")
else:
    print("Invalid PIN")
```
