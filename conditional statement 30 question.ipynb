```python
# find the greatest amond three numbers using nested if-else
a=int(input("Enter first number:"))
b=int(input("Enter second number:"))
c=int(input("Enter third number:"))
if a>b:
    if a>c:
        print("Greatest=",a)
    else:
        print("Greatest=",c)
else:
    if b>c:
        print("Greatest=",b)
    else:
        print("Greatest=",c)
```
